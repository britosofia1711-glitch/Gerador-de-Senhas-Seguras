```python
import json
import os

# Começa com a lista vazia e define a constante para o nome do arquivo
livros = []
NOME_ARQUIVO = "biblioteca.json"


def carregar_dados():
  """Lê o arquivo ao iniciar o programa para recuperar os livros salvos."""
  global livros
  if os.path.exists(NOME_ARQUIVO):
    try:
      with open(NOME_ARQUIVO, "r", encoding="utf-8") as f:
        livros = json.load(f)
    except Exception as e:
      print(f"Erro ao carregar os dados: {e}")
      livros = []


def salvar_dados():
  """Salva a lista atual de livros no arquivo."""
  try:
    with open(NOME_ARQUIVO, "w", encoding="utf-8") as f:
      json.dump(livros, f, ensure_ascii=False, indent=4)
  except Exception as e:
    print(f"Erro ao salvar os dados: {e}")


def exibir_menu():
  """Exibe o menu principal do sistema."""
  print("\n===== SISTEMA DE BIBLIOTECA =====")
  print("1. Cadastrar Livro")
  print("2. Listar Livros")
  print("3. Buscar Livro (Título ou Autor)")
  print("4. Registrar Empréstimo")
  print("5. Registrar Devolução")
  print("6. Estatísticas do Acervo (Extra)")
  print("0. Sair")


def cadastrar_livro():
  """Registra um novo livro solicitando título, autor e ano."""
  print("\n--- Cadastro de Novo Livro ---")
  titulo = input("Digite o título do livro: ").strip()
  autor = input("Digite o autor do livro: ").strip()

  # Primeiro tenta converter o ano digitado num número inteiro e depois detecta o erro se o usuário digitar letras ou símbolos errados
  try:
    ano = int(input("Digite o ano de publicação: "))
  except ValueError:
    print("Erro: O ano deve ser um número inteiro válido. Livro não cadastrado.")
    return

  livro = {"titulo": titulo, "autor": autor, "ano": ano, "disponivel": "Sim"}
  livros.append(livro)
  salvar_dados()
  print(f"Sucesso! O livro '{titulo}' foi cadastrado.")


# --- PROGRAMA PRINCIPAL ---
# Carrega os dados salvos e inicia o loop do menu
carregar_dados()

while True:
  exibir_menu()
  opcao = input("Escolha uma opção: ").strip()

  if opcao == "1":
    cadastrar_livro()
  elif opcao == "0":
    print("Saindo do programa... Até logo!")
    break
  else:
    print(
        "Opção inválida ou ainda em desenvolvimento! Por favor, escolha 1 ou 0."
    )
```
